```python
# fermentation_simulation.py
# Complete Python script for dynamic aerobic fermentation simulation

import numpy as np
import matplotlib.pyplot as plt

# === PARAMETERS ===
# Yield coefficients and constants
Y_O2_X = 1.1     # g O2 / g biomass
Y_X_S = 0.5      # g biomass / g glucose
Y_P_S = 0.05     # g ethanol / g glucose
Y_CO2_S = 1.0    # g CO2 / g glucose
r_S = 2          # g glucose / L*h
mu = 0.3         # 1/h
k_La = 50        # 1/h
P_O2 = 0.21      # atm
H = 769.2        # atm·L/mol
C_O2 = 0.005     # g/L
epsilon_g = 0.15
C_O2_min = 0.002

# Initial conditions
S0 = 50          # g/L
X0 = 0.1         # g/L
pH_0 = 5.5
pH_min = 5.0
T_0 = 30         # °C
T_max = 35

# Other constants
rho_g = 1.225    # kg/m³
g = 9.81         # m/s²
H_t = 0.3        # m
eta = 0.5
Y_Q_X = 15       # kJ/g biomass
Cp = 4.18        # kJ/kg·K
m = 0.5          # kg
k_CO2 = 0.05     # pH drop per g CO2
k_f = 0.02       # L foam / g gas
Foam_threshold = 0.1
F_O2 = 0.001785 / 60000   # m³/s

# Time scale
time = np.arange(0, 49, 1)

# === INITIALIZE ARRAYS ===
S_t = np.zeros_like(time, dtype=float)
X_t = np.zeros_like(time, dtype=float)
CO2_t = np.zeros_like(time, dtype=float)
OUR_t = np.zeros_like(time, dtype=float)
OTR_t = np.zeros_like(time, dtype=float)
F_O2_t = np.zeros_like(time, dtype=float)
pH_t = np.zeros_like(time, dtype=float)
NaOH_added = np.zeros_like(time, dtype=float)
Temp_t = np.zeros_like(time, dtype=float)
Foam_t = np.zeros_like(time, dtype=float)
AntiFoam_added = np.zeros_like(time, dtype=float)

# === SET INITIAL CONDITIONS ===
S_t[0] = S0
X_t[0] = X0
pH_t[0] = pH_0
Temp_t[0] = T_0
F_O2_t[0] = F_O2

# === SIMULATION LOOP ===
for t in range(1, len(time)):
    X_t[t] = X_t[t-1] * np.exp(mu)
    S_t[t] = S_t[t-1] * np.exp(-r_S)
    CO2_t[t] = Y_CO2_S * (S0 - S_t[t])
    OUR_t[t] = Y_O2_X * X_t[t]
    OTR_t[t] = k_La * (P_O2 / H - C_O2)

    # Oxygen flow adjustment
    if OUR_t[t] > OTR_t[t]:
        F_O2_t[t] = min(F_O2_t[t-1] * 1.1, 0.01)
    elif C_O2 < C_O2_min:
        F_O2_t[t] = min(F_O2_t[t-1] * 1.2, 0.01)
    else:
        F_O2_t[t] = F_O2_t[t-1]

    # pH control
    pH_t[t] = pH_0 - k_CO2 * CO2_t[t]
    if pH_t[t] < pH_min:
        NaOH_added[t] = (pH_min - pH_t[t]) * 2
        pH_t[t] = pH_min

    # Temperature control
    heat = Y_Q_X * X_t[t]
    Temp_t[t] = T_0 + heat / (Cp * m)
    if Temp_t[t] > T_max:
        Temp_t[t] = T_max

    # Foam control
    foam = k_f * (CO2_t[t] + Y_P_S * (S0 - S_t[t]))
    if foam > Foam_threshold:
        Foam_t[t] = Foam_threshold
        AntiFoam_added[t] = 1
    else:
        Foam_t[t] = foam

# === PLOT RESULTS ===
plt.figure(figsize=(12, 10))

plt.subplot(3, 2, 1)
plt.plot(time, S_t, 'b')
plt.title('Glucose Depletion')
plt.xlabel('Time (h)'); plt.ylabel('Glucose (g/L)')

plt.subplot(3, 2, 2)
plt.plot(time, X_t, 'r')
plt.title('Biomass Growth')
plt.xlabel('Time (h)'); plt.ylabel('Biomass (g/L)')

plt.subplot(3, 2, 3)
plt.plot(time, pH_t, 'g')
plt.title('pH Variation')
plt.xlabel('Time (h)'); plt.ylabel('pH')

plt.subplot(3, 2, 4)
plt.plot(time, Temp_t, 'orange')
plt.title('Temperature Rise')
plt.xlabel('Time (h)'); plt.ylabel('Temperature (°C)')

plt.subplot(3, 2, 5)
plt.plot(time, Foam_t, 'purple')
plt.title('Foam Volume')
plt.xlabel('Time (h)'); plt.ylabel('Foam (L)')

plt.subplot(3, 2, 6)
plt.plot(time, F_O2_t * 60000, 'k')
plt.title('Oxygen Flow Rate')
plt.xlabel('Time (h)'); plt.ylabel('Flow (L/min)')

plt.tight_layout()
plt.show()

# === SUMMARY OUTPUT ===
print(f"Final Temperature: {Temp_t[-1]:.2f} °C")
print(f"Total NaOH Added: {np.sum(NaOH_added):.2f} g/L")
print(f"Total Anti-Foam Used: {np.sum(AntiFoam_added):.2f} mL")
```
